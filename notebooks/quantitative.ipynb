## Filling a feature

A window is searched over one whole feature at a time, so the feature is the unit
everything below is measured on. This section focuses on what a feature is ideally
expected to cover.

It's important to note that features differ enormously in size, so the width used here is
the middling one rather than the average, which a handful of very wide features would
otherwise stand for.

### What it takes to fill a feature

How much of an instrument a whole feature of the middling width would swallow.

It's important to note that SHARAD is consider as a line made up of traces.

```python
"""Tabulate how much of each instrument it takes to cover a whole feature."""

filling.filling(read)
```

### What each instrument lands on a feature

## Coverage

Now that what a feature holds has been described, we can look at what the filter would have kept and thrown away.

### How much of a feature each instrument reaches

What the chosen window actually leaves on the ground, counting a cell once
however often it was revisited. Only features that earned a window are counted, and
an instrument absent from one counts as nothing there rather than being left out.

```python
"""Tabulate how much of a feature each instrument reaches and how much they share."""

coverage.reached(read)
```

### How long a window runs

**Time Window Score** is the geometric mean of the shares the window's own
  constraints reach, so as the filter is written now it is exactly
  `sqrt(CTX share of the feature x CRISM share of the feature)`. SHARAD is `timeless`,
  so it is asked of the whole record rather than of the window and never enters
  the score. The days a window runs are priced against the ground it reaches
  while the search is choosing the window, but that charge is not carried into
  the number reported here.

```python
"""Tabulate how long a window runs and how far it reaches."""

coverage.windows(read)
```
